**1 - Configuration yaml**

```python
# Paths and parameters
roi_path = "/content/drive/MyDrive/roi/roi.gpkg"
start_date = "2023-07-30"
end_date = "2023-08-30"
id_feature_swot = 2320106762
classes_to_keep = [1, 2, 3, 4]
output_path = "/content/drive/MyDrive/output"
yaml_path = "/content/drive/MyDrive/config.yaml"
vt_path = "/content/drive/MyDrive/vt.csv"
template_global_yaml_path = (
    "/content/drive/MyDrive/implicity-bathy/example_config/config_all.yaml"
)
```

```python
import yaml

# Read template files
with open(template_global_yaml_path, "r") as f:
    config = yaml.safe_load(f)

# Write parameters
config["dataprocessing"]["output_path"] = output_path
config["dataprocessing"]["roi_gpkg"] = roi_path
config["dataprocessing"]["start_date"] = start_date
config["dataprocessing"]["end_date"] = end_date
config["dataprocessing"]["id_feature_swot"] = id_feature_swot
config["inference"]["vt_path"] = vt_path
config = {k: ("" if v is None else v) for k, v in config.items()}

with open(yaml_path, "w") as f:
    yaml.safe_dump(config, f, sort_keys=False)
```

**2 - Creating folders**



```python
from pathlib import Path


def create_folder(output_path):
    if not Path(output_path).exists():
        Path(output_path).mkdir()
        list_folder = [
            "cfg",
            "data",
            "dsm",
            "inference",
            "inputs_nc",
            "logs",
            "train",
        ]
        for name in list_folder:
            folder = Path(output_path) / name.strip()
            if not Path(folder).exists():
                Path(folder).mkdir(parents=True)


create_folder(output_path)
```
